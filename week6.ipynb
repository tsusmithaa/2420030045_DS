```python
import pandas as pd 
from sklearn.datasets import load_iris

# Load data
iris = load_iris()
df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
print(df.head())

# Mean
mean_values = df.mean()
print("\nMean values of each feature:")
print(mean_values)

# Median
median_values = df.median()
print("\nMedian values of each feature:")
print(median_values)

# Mode
mode_values = df.mode().iloc[0]
print("\nMode values of each feature:")
print(mode_values)
```

```python
import pandas as pd 
import seaborn as sns

# Load data
tips = sns.load_dataset('tips')
df = pd.DataFrame(data=tips)
print(df.head())

#mean
mean_values = df.mean(numeric_only=True)
print("\nMean values of each feature:")
print(mean_values)

# Median 
median_values = df.median(numeric_only=True)
print("\nMedian values of each feature:")
print(median_values)

# Mode 
mode_values = df.mode().iloc[0]
print("\nMode values of each feature:")
print(mode_values)
```

```python
import pandas as pd
from sklearn.datasets import load_iris

iris = load_iris()
df=pd.DataFrame(data=iris.data, columns=iris.feature_names)
print(df.head())

#range(max-min)
range_values = df.max() - df.min()
print(range_values)

std_values = df.std()
print("\nStandard Deviation values of each feature:")
print(std_values)

#interpercentile
Q1 = df.quantile(0.25)
Q3 = df.quantile(0.75)
IQR = Q3 - Q1
print("\nInterquartile Range (IQR) values of each feature:")
print(IQR)
```

```python
import pandas as pd
import seaborn as sns

# Load data
tips = sns.load_dataset('tips')
df = pd.DataFrame(data=tips)
print(df.head())

numeric_df = df.select_dtypes(include='number')

# 1. Range (max - min)
range_values = numeric_df.max() - numeric_df.min()
print("\nRange (Max - Min) values of each feature:")
print(range_values)

# 2. Standard Deviation
std_values = numeric_df.std()
print("\nStandard Deviation values of each feature:")
print(std_values)

# 3. Interquartile Range (IQR)
Q1 = numeric_df.quantile(0.25)
Q3 = numeric_df.quantile(0.75)
IQR = Q3 - Q1
print("\nInterquartile Range (IQR) values of each feature:")
print(IQR)
```

```python
import pandas as pd
from sklearn.datasets import load_iris

iris = load_iris()
df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
print(df.head())

#skewness
skewness_values = df.skew()
print("\nSkewness values of each feature:")
print(skewness_values)

#kurtosis
kurtosis_values = df.kurtosis()
print("\nKurtosis values of each feature:")
print(kurtosis_values)

```
