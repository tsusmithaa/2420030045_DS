```python
from statistics import kde

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import  load_iris

iris = load_iris()
df=pd.DataFrame(data=iris.data, columns=iris.feature_names)
df['species'] = iris.target


#univaritate visualization
#1 Histogram
plt.figure(figsize=(10,6))
sns.histplot(df['sepal length (cm)'], bins=20, kde=True)
plt.title('Histogram of Sepal Length')
plt.show()

#boxplot
plt.figure(figsize=(6,4))
sns.boxplot(df['sepal length (cm)'])
plt.title('Boxplot of Sepal Length')
plt.show()

#pie chart
species_counts = df['species'].value_counts()
plt.figure(figsize=(8,8))
plt.pie(species_counts, labels=species_counts.index, autopct='%1.1f%%', startangle=90)
plt.title('Pie Chart of Species Distribution')  
plt.show()
```

```python
#2 Bivariate visualization

#scatter plot
plt.figure(figsize=(10,6))
sns.scatterplot(x='sepal length (cm)', y='sepal width (cm)', hue='species', data=df)
plt.title('Scatter Plot of Sepal Length vs Sepal Width')
plt.show()

#line chart (trend of sepal length across samples )
plt.figure(figsize=(10,6))
plt.plot(df['sepal length (cm)'])
plt.title('Line Chart of Sepal Length Across Samples')
plt.xlabel('Sample Index')
plt.ylabel('Sepal Length (cm)')
plt.show()

#bar chart 
plt.figure(figsize=(6,4))
sns.barplot(x=df['species'], y=df['sepal length (cm)'])
plt.title('Bar Chart of Sepal Length by Species')
plt.xticks(ticks=[0, 1, 2], labels=iris.target_names)   
plt.show()
```

```python
#multivariate visualization

#heatmap
plt.figure(figsize=(10,8))
sns.heatmap(df.iloc[:, :-1].corr(), annot=True, cmap='coolwarm')
plt.title('Heatmap of Feature Correlations')
plt.show()

#bubble chart 
plt.figure(figsize=(10,6))
plt.scatter(df['sepal length (cm)'], df['sepal width (cm)'], s=df['petal length (cm)']*20, c=df['species'], cmap='viridis', alpha=0.6)
plt.title('Bubble Chart of Sepal Length vs Sepal Width with Petal Length as Size')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Sepal Width (cm)')
plt.show()


#pairplot 
sns.pairplot(df.iloc[:,:-1], hue='species', palette='bright')
plt.suptitile('Pairplot of Iris Dataset Features',y=1.02)
plt.show()

```

```python
#4 advanced visualization
#violin plot
plt.figure(figsize=(10,6))
sns.violinplot(x='species', y='sepal length (cm)', data=df, palette='pastel')
plt.title('Violin Plot of Sepal Length by Species')
plt.xticks(ticks=[0, 1, 2], labels=iris.target_names)
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load Tips dataset
tips = sns.load_dataset('tips')

# Create DataFrame
df = pd.DataFrame(tips)

# Histogram
plt.figure(figsize=(6, 4))
sns.histplot(df['total_bill'], bins=20, kde=True)
plt.title('Distribution of Total Bill')
plt.show()

# Boxplot
plt.figure(figsize=(6, 4))
sns.boxplot(x=df['total_bill'])
plt.title('Boxplot of Total Bill')
plt.show()

# Count the number of customers for each day
day_counts = df['day'].value_counts()

# Pie chart
plt.figure(figsize=(8, 8))
plt.pie(day_counts, labels=day_counts.index, autopct='%1.1f%%')
plt.title('Distribution by Day')
plt.show()
```

```python
# #2 Bivariate visualization

# scatter plot
plt.figure(figsize=(10,6))
sns.scatterplot(x='total_bill', y='tip', hue='sex', data=tips)
plt.title('Scatter Plot of Total Bill vs Tip')
plt.show()

# line chart (trend of total bill across samples)
plt.figure(figsize=(10,6))
plt.plot(tips['total_bill'])
plt.title('Line Chart of Total Bill Across Samples')
plt.xlabel('Sample Index')
plt.ylabel('Total Bill')
plt.show()

# bar chart
plt.figure(figsize=(6,4))
sns.barplot(x='day', y='total_bill', data=tips)
plt.title('Bar Chart of Total Bill by Day')
plt.xlabel('Day')
plt.ylabel('Total Bill')
plt.show()
```

```python
# multivariate visualization

# heatmap
plt.figure(figsize=(10,8))

sns.heatmap(df.select_dtypes(include='number').corr(), annot=True, cmap='coolwarm')

plt.title('Heatmap of Tips Dataset Feature Correlations')

plt.show()


# bubble chart

plt.figure(figsize=(10,6))

plt.scatter(
    df['total_bill'],
    df['tip'],
    s=df['size'] * 50,
    c=df['total_bill'],
    cmap='viridis',
    alpha=0.6
)

plt.title('Bubble Chart of Total Bill vs Tip with Party Size as Bubble Size')

plt.xlabel('Total Bill')

plt.ylabel('Tip')

plt.show()


# pairplot

sns.pairplot(
    df[['total_bill', 'tip', 'size', 'sex']],
    hue='sex',
    palette='bright'
)

plt.suptitle('Pairplot of Tips Dataset Features', y=1.02)

plt.show()
```
